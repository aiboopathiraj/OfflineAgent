### **RAG - LLama Index**

```python
import kagglehub
path = kagglehub.dataset_download("timoboz/data-science-cheat-sheets")
```

```python
import os
dir_path = []
for i in os.listdir(path):
  main_ = path+"/"+i
  if os.path.isdir(main_):
    for p in os.listdir(main_):
      if p.endswith(".pdf"):
          file = os.path.join(main_,p)
          if os.path.exists(file):
            dir_path.append(file)
```

```python
dir_path
```

```python
!uv pip install pymupdf llama-index-readers-file
```

```python
try:
  from llama_index.core import SimpleDirectoryReader
except:
  from llama_index.readers import SimpleDirectoryReader
from llama_index.readers.file import PyMuPDFReader

# Force LlamaIndex to use the local PyMuPDF parser without API calls
file_extractor = {".pdf": PyMuPDFReader()}
documents = SimpleDirectoryReader(
    input_files=dir_path,
    recursive=True,
    required_exts=[".pdf", ".md", ".docx", ".txt"],
    file_extractor=file_extractor
).load_data()

print(f"Loaded {len(documents)} documents")
```

```python
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

embedding = HuggingFaceEmbedding(model_name='sentence-transformers/all-MiniLM-L6-v2',embed_batch_size=64)
```

```python
import psycopg2
import textwrap

connection_string = "postgresql://postgres:password@localhost:5432"
db_name = "vector_db"
conn = psycopg2.connect(connection_string)
conn.autocommit = True

with conn.cursor() as c:
    # Terminate all other active connections to the target database
    c.execute(f"""
        SELECT pg_terminate_backend(pg_stat_activity.pid)
        FROM pg_stat_activity
        WHERE pg_stat_activity.datname = '{db_name}'
          AND pid <> pg_backend_pid();
    """)
    c.execute(f"DROP DATABASE IF EXISTS {db_name}")
    c.execute(f"CREATE DATABASE {db_name}")

print(f"Database '{db_name}' successfully recreated!")
```

```python
import sys
import sqlalchemy.util
import sqlalchemy.util.concurrency

# 1. Patch Python version flag
setattr(sqlalchemy.util, 'py310', True)

# 2. Patch TypedDict
import sqlalchemy.util.typing
if not hasattr(sqlalchemy.util.typing, 'TypedDict'):
    from typing import TypedDict
    setattr(sqlalchemy.util.typing, 'TypedDict', TypedDict)

# 3. Patch missing async/concurrency helpers
if not hasattr(sqlalchemy.util.concurrency, 'await_fallback'):
    def await_fallback(coroutine):
        return coroutine
    def await_only(coroutine):
        return coroutine
    def in_greenlet():
        return False

    setattr(sqlalchemy.util.concurrency, 'await_fallback', await_fallback)
    setattr(sqlalchemy.util.concurrency, 'await_only', await_only)
    setattr(sqlalchemy.util.concurrency, 'in_greenlet', in_greenlet)

# Now proceed with imports
from sqlalchemy import make_url
from llama_index.vector_stores.postgres import PGVectorStore
from llama_index.core import StorageContext, VectorStoreIndex

# Clean both documents and nodes of NUL (0x00) characters to prevent PostgreSQL insertion errors
for doc in documents:
    if doc.text:
        doc.set_content(doc.text.replace('\x00', ''))
    for key, val in list(doc.metadata.items()):
        if isinstance(val, str):
            doc.metadata[key] = val.replace('\x00', '')

for node in nodes:
    if node.text:
        node.set_content(node.text.replace('\x00', ''))
    for key, val in list(node.metadata.items()):
        if isinstance(val, str):
            node.metadata[key] = val.replace('\x00', '')

url = make_url(connection_string)
vector_store = PGVectorStore.from_params(
    database=db_name,
    host=url.host,
    password=url.password,
    port=url.port,
    user=url.username,
    table_name="study_materials",
    embed_dim=384,  # Matching the 384 dimensions of sentence-transformers/all-MiniLM-L6-v2
    hnsw_kwargs={
        "hnsw_m": 16,
        "hnsw_ef_construction": 64,
        "hnsw_ef_search": 40,
        "hnsw_dist_method": "vector_cosine_ops",
    },
)

storage_context = StorageContext.from_defaults(vector_store=vector_store)
index = VectorStoreIndex.from_documents(
    documents, storage_context=storage_context, show_progress=True, embed_model=embedding, insert_batch_size=2048,
)
```
